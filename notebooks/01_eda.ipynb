```python
converting_total_charges_to_right_dtype = pd.to_numeric(df["TotalCharges"], errors="coerce").isna()
converting_total_charges_to_right_dtype
```

```python
bad_total_charge_rows = df[converting_total_charges_to_right_dtype]
print(f"Number of bad TotalCharges rows: {len(bad_total_charge_rows)}")
print("\n")
print(bad_total_charge_rows[["customerID", "tenure", "MonthlyCharges", "TotalCharges"]])
```

```python
df_clean = df.copy()
print(f"Original data shape: {df.shape}")
print(f"Clean data shape: {df_clean.shape}")
```

```python
print("Before fixing TotalCharges data type:", df_clean["TotalCharges"].dtypes)
df_clean["TotalCharges"] = pd.to_numeric(df_clean["TotalCharges"], errors="coerce")
print("After fixing TotalCharges data type:", df_clean["TotalCharges"].dtypes)
print(f"Total missing value of TotalCharges: {df_clean["TotalCharges"].isnull().sum()}")
```

```python
df_clean.isnull().sum()
```

```python
all_rows_TotalCharges = df_clean["TotalCharges"].isnull()                                                          

missing_rows_TotalCharges = df_clean[all_rows_TotalCharges]

missing_rows_TotalCharges[["customerID", "tenure", "MonthlyCharges", "TotalCharges"]] 
```

```python
df_clean.loc[all_rows_TotalCharges, "TotalCharges"] = 0

missing_rows_TotalCharges[["customerID", "tenure", "MonthlyCharges", "TotalCharges"]] 
```
